#Clean text, remove empty/duplicate rows

```python
import re
import html
from pythainlp.tokenize import word_tokenize

# log: how many rows after each step
log = []
def log_step(name, data):
    log.append({
        "step": name,
        "rows": len(data),
        "label_1": int((data["label"] == 1).sum()),   # suicidal ideation
        "label_0": int((data["label"] == 0).sum()),   # not
    })

# load fresh and keep only what we need
df = pd.read_csv("SIED-Thai.csv", encoding="utf-8")
df = df.rename(columns={"Tweets": "tweet", "Labels": "label"})   # Kept the same names as in Arabic
df = df.drop(columns=["No."])  # old row number, not needed
log_step("0. raw", df)


# 1. remove the empty tweets

df = df.dropna(subset=["tweet"])
df["tweet"] = df["tweet"].astype(str)
df = df[df["tweet"].str.strip() != ""]
log_step("1. drop empty", df)


# 2. clean the text

REMOVE_HASHTAGS = True

def clean_text(t):
    t = t.replace("_x000D_", " ")                                      # Excel line-break code
    t = html.unescape(t)                                               # &amp; -> &
    t = re.sub(r"[\u200b\u200e\u200f\u202a-\u202e\u2066-\u2069]", "", t)  # invisible characters
    t = re.sub(r"^RT\s+", "", t)                                       # retweet marker
    t = re.sub(r"https?://\S+|www\.\S+", "", t)                        # links
    t = re.sub(r"@\w+", "[USER]", t)                                   # usernames - [USER] (privacy)
    if REMOVE_HASHTAGS:
        t = re.sub(r"#\S+", "", t)                                     # remove hashtags (# + everything up to the next space)
    t = re.sub(r"(\.\s*){3,}", "... ", t)
    t = re.sub(r"_{3,}", " ", t)                                       # separator lines
    t = re.sub(r"\s+", " ", t).strip()                                 # extra spaces / line breaks
    return t

df["tweet"] = df["tweet"].apply(clean_text)
df = df[df["tweet"] != ""]          # tweets that were only hashtags/links are now empty so remove them
log_step("2. clean text", df)

#3. remove contradictory tweets:

label_counts = df.groupby("tweet")["label"].nunique()
conflicting = label_counts[label_counts > 1].index
df = df[~df["tweet"].isin(conflicting)]
log_step("3. drop conflicting", df)

#4. remove exact duplicates:

df = df.drop_duplicates(subset="tweet")
log_step("4. drop duplicates", df)

print(pd.DataFrame(log))

#count Thai words (PyThaiNLP), now that hashtags are gon
def count_thai_words(text):
    words = word_tokenize(text, engine="newmm")
    return len([w for w in words if w.strip() != ""])   # don't count spaces as words

df["n_words"] = df["tweet"].apply(count_thai_words)

print("\nWords per tweet, by label:")
print(df.groupby("label")["n_words"].describe())

print("\nHow many tweets per label would survive a minimum length:")

for m in [10, 15, 20, 30]:

    kept = df[df["n_words"] >= m]["label"].value_counts()
    print(f"  >= {m} words ->  label 1: {kept.get(1,0)},  label 0: {kept.get(0,0)}")
```

#Keep long-enough tweets, balanced sample, save

```python
MIN_WORDS = 15      # same cutoff as Arabic, for a fair comparison
N_PER_CLASS = 50
SEED = 42           # same seed as Arabic

# Save all cleaned tweets before the length filter (for the length histogram)
df.to_csv("thai_clean_before_length_filter.csv", index=False, encoding="utf-8-sig")

# 1. keep tweets with at least 15 words

df_long = df[df["n_words"] >= MIN_WORDS].copy()
log_step(f"5. keep >= {MIN_WORDS} words", df_long)

# 2. mark tweets containing "ซึมเศร้า" (depression) - just a flag, nothing removed

df_long["has_keyword"] = df_long["tweet"].str.contains("ซึมเศร้า")

# 3. random 50 from label 1 + 50 from label 0, then shuffle

sample = pd.concat([
    df_long[df_long["label"] == 1].sample(n=N_PER_CLASS, random_state=SEED),
    df_long[df_long["label"] == 0].sample(n=N_PER_CLASS, random_state=SEED),
])

sample = sample.sample(frac=1, random_state=SEED).reset_index(drop=True)

sample.insert(0, "id", [f"th_{i+1:03d}" for i in range(len(sample))])
sample.insert(1, "language", "Thai")
log_step("6. balanced sample", sample)

# 4. check
print(pd.DataFrame(log))
print("\nKeyword 'ซึมเศร้า' in sample (by label):")
print(pd.crosstab(sample["label"], sample["has_keyword"]))

# 5. save
df_long.to_csv("thai_clean_full.csv", index=False, encoding="utf-8-sig")
pd.DataFrame(log).to_csv("thai_cleaning_log.csv", index=False)
print("\nSaved.")
```

#Split each tweet into 3 equal parts:

```python
def split_into_thirds_thai(text):
    # Split into Thai words, keeping the spaces as their own pieces, so we can glue everything back exactly as it was
    tokens = word_tokenize(text, engine="newmm", keep_whitespace=True)

    # Positions of the real words (not spaces)

    real = [i for i, tk in enumerate(tokens) if tk.strip() != ""]
    n = len(real)

    # Same sizing rule as Arabic: leftover words go to the first part(s)
    base, extra = n // 3, n % 3
    sizes = [base + (1 if i < extra else 0) for i in range(3)]

    cut1 = real[sizes[0]]
    cut2 = real[sizes[0] + sizes[1]]

    # Glue tokens back together
    prefix = "".join(tokens[:cut1]).strip()
    middle = "".join(tokens[cut1:cut2]).strip()
    suffix = "".join(tokens[cut2:]).strip()

    return prefix, middle, suffix

sample[["prefix", "middle", "suffix"]] = sample["tweet"].apply(
    lambda t: pd.Series(split_into_thirds_thai(t))
)
sample = sample.rename(columns={"tweet": "full"})

# Safety check: the 3 parts glued together = the full tweet (ignoring spaces)
rebuilt = (sample["prefix"] + sample["middle"] + sample["suffix"]).str.replace(" ", "")
print("All splits correct:", (rebuilt == sample["full"].str.replace(" ", "")).all())

# Where does "ซึมเศร้า" appear in the suicidal (label 1) tweets?
kw = "ซึมเศร้า"
for part in ["prefix", "middle", "suffix"]:
    sample[f"kw_in_{part}"] = sample[part].str.contains(kw)
print("\nKeyword position (label 1 only):")
print(sample[sample["label"] == 1][["kw_in_prefix", "kw_in_middle", "kw_in_suffix"]].sum())

# Neat column order (+ keep the emotion columns, they may be useful later)
cols = ["id", "language", "label", "full", "prefix", "middle", "suffix",
        "n_words", "has_keyword", "kw_in_prefix", "kw_in_middle", "kw_in_suffix",
        "Happiness", "Sadness", "Anger", "Disgust", "Surprise", "Fear", "Sentiment"]
sample = sample[cols]

sample.to_csv("thai_sample_split.csv", index=False, encoding="utf-8-sig")
print("\nSaved thai_sample_split.csv")
sample[["id", "label", "full", "prefix", "middle", "suffix"]].head(5)
```
